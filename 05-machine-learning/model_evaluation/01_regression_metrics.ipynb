<h1> Model Evaluation: Regression Metrics </h3>

```python
import numpy as np
```

<h3> Metric 1 — MAE </h3>

Average actual and predicted distance.

```python
def mean_absolute_error(
    y_true,
    y_pred
):

    return np.mean(
        np.abs(
            y_true - y_pred
        )
    )
```

<h3> Metric 2 — MSE </h3>

Difference from MAE: Here, the error is squared.

MSE is differentiable.

Absolute value (MAE):

$$ |x| $$

does not have a defined derivative at zero.

However:

$$ x^2 $$

is perfectly smooth.

Therefore, Gradient Descent is easier to perform.

```python
def mean_squared_error(
    y_true,
    y_pred
):

    return np.mean(
        (
            y_true-y_pred
        )**2
    )
```

<h3> Metric 3 — RMSE </h3>

The issue with MSE:

Its unit changes.

Target:

For example: Million dollars

MSE: Million dollars squared

So:</br>
We take the square root:

$$ RMSE= \sqrt{MSE} $$

```python
def root_mean_squared_error(
    y_true,
    y_pred
):

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    return np.sqrt(mse)
```

<h3> Metric 4 — R² Score </h3>

This is the most important metric in regression.

Question:

How much of the variation in the target variable does the model explain?

Formula:

$$ R^2= 1- \frac{SS_{res}} {SS_{tot}} $$


1. Part One:

Residual:

$$ SS_{res} = \sum(y-\hat y)^2 $$

Meaning:

Model error.


2. Part Two:

Total:

$$ SS_{tot} = \sum(y-\bar y)^2 $$

That is:

Variation in the data itself.

For example:

If:

$$ R^2=0.8 $$

It means:

The model has been able to explain 80% of the variation in house prices.

```python
def r2_score(
    y_true,
    y_pred
):

    ss_res = np.sum(
        (y_true-y_pred)**2
    )

    ss_tot = np.sum(
        (y_true-np.mean(y_true))**2
    )

    return 1 - (
        ss_res / ss_tot
    )
```
