```python
from src.games import TicTacToe

from src.alphabeta import (
    alpha_beta_value,
    alpha_beta_decision,
    alpha_beta_value_with_stats,
    alpha_beta_decision_ordered,
    alpha_beta_value_ordered_with_stats,
)
from src.metrics import (
    SearchStats
)
from src.minimax import (
    minimax_value,
)
```

Alpha-Beta says: </br>
If a part of the tree has no other impact on the final decision, don't examine it.

That is:</br>
Pruning = removing unnecessary branches

**Alpha (α):** </br>
The best guaranteed value for MAX so far.</br>
That is: The minimum value MAX can take is this.

**Beta (β):**</br>
The best guaranteed value for MIN.</br>
That is: The maximum value MIN allows is this.

Alpha-Beta = Minimax + Pruning

```python
board1 = (
    "X", "X", " ",
    "O", "O", " ",
    " ", " ", " ",
)

board2 = (
    "X", " ", " ",
    "O", "O", " ",
    "X", " ", " ",
)

board3 = (
    "O", "O", " ",
    "X", "X", " ",
    " ", " ", " ",
)
```

```python
game = TicTacToe()

game
```

```python
state = (
    board3,
    "X",
)
```

alpha beta value test on a winning State → value 10000

```python
value = alpha_beta_value(
    game,
    state,
    alpha=-float("inf"),
    beta=float("inf"),
)

value
```

alpha beta decision test on a specific State → Valid Action

```python
action = alpha_beta_decision(
    game,
    state,
)

action
```

alpha beta decision test on initial State (Game start test) → Valid Action

Why does the order of actions affect Alpha-Beta? </br>
Because Alpha-Beta will prune more if it sees good moves earlier.

<h3> Step 6.1: Measuring Alpha-Beta Pruning Effect </h3>

But the main purpose of Alpha-Beta: </br>
Reducing the number of checked nodes without changing is the answer.

```python
game = TicTacToe()

state = game.initial_state()
```

```python
stats = SearchStats()

value = alpha_beta_value_with_stats(
    game,
    state,
    alpha=-float("inf"),
    beta=float("inf"),
    stats=stats,
)

stats.summary()
```

<h3> Step 6.2: Comparing with complete min-max </h3>

**complete Minimax** </br>
Nodes Visited:     549945</br>
Terminal States:   255168</br>
Max Depth:         9


**Alpha-Beta**</br>
Nodes Visited:     18297</br>
Terminal States:   7330</br>
Max Depth:         0

*About 96.6% reduction in search space*

```python
minimax_value_result = minimax_value(
    game,
    state,
)

alpha_beta_value_result = alpha_beta_value(
    game,
    state,
    alpha=-float("inf"),
    beta=float("inf"),
)


print(minimax_value_result)
print(alpha_beta_value_result)
```

In Tic-Tac-Toe from the starting position: </br>

If both players play optimally:</br>
X cannot force a win.</br>
O cannot force a win either.

The optimal outcome of the game is: Draw.

And in our utility:</br>
Draw = 0

```python
state = (
    board3,
    "X",
)

minimax_value_result = minimax_value(
    game,
    state,
)

alpha_beta_value_result = alpha_beta_value(
    game,
    state,
    alpha=-float("inf"),
    beta=float("inf"),
)

print(minimax_value_result)
print(alpha_beta_value_result)
```

We checked:

*Minimax(s)=AlphaBeta(s)*

That is:</br>
Alpha-Beta only reduces Search, not the result.

Move Ordering is a concept within Alpha-Beta. </br>
That is, it is an optimization, not a new algorithm.

Alpha-Beta does the most pruning when:</br>
It checks good moves early.
