<h1> Probability and Bayesian Reasoning

Probability provides a mathematical framework for reasoning under uncertainty.

Unlike deterministic systems where the output is completely known, many real-world AI problems involve uncertainty:

- Medical diagnosis
- Weather prediction
- Spam detection
- Robotics decision making
- Natural language processing

Probabilistic reasoning allows AI systems to represent uncertainty and update beliefs based on new evidence.

### Random Variables

A random variable maps outcomes of a random experiment to numerical values.

Two main types:

#### Discrete Random Variables

Have countable values.

Examples:

- Dice rolls
- Number of customers
- Number of failures


#### Continuous Random Variables

Can take any value in an interval.

Examples:

- Height
- Temperature
- Time

Formula:

$P(A|B)=\frac{P(A \cap B)}{P(B)}$

Meaning:
The probability of A given that B has occurred.

Visualization

```python
priors = np.linspace(
    0.01,
    0.5,
    50
)

posteriors=[]

for p in priors:
    posterior = (
        0.9*p
        /
        (
            0.9*p
            +
            0.05*(1-p)
        )
    )

    posteriors.append(
        posterior
    )

plt.figure(
    figsize=(8,5)
)

plt.plot(
    priors,
    posteriors
)

plt.xlabel(
    "Prior Probability"
)

plt.ylabel(
    "Posterior Probability"
)

plt.title(
    "Bayesian Update"
)

plt.grid()
plt.show()
```

This chart illustrates the core concept of Bayesian updating precisely.

In Bayesian reasoning, we start with an initial belief and then update our probability as we receive new information.

### The Prior Probability Effect

On the left side of the graph:
Prior Probability ≈ 0.01

The initial probability is low.
However, after observing the evidence, the posterior value increases:

Prior ≈ 0.01

Posterior ≈ 0.15

This means that even if the initial belief is low, strong evidence can increase the probability.

### Increasing the prior leads to an increase in the posterior.

Moving along the x-axis:</br>
Prior Probability ↑

We see that:</br>
Posterior Probability ↑

...increases.

In other words, if the initial probability of an event occurring is higher, the final belief—after observing the same evidence—will also be higher.

### Non-linear behavior of the plot

The plot is not linear. The change in the posterior is rapid.</br>
However, at higher values, the increase in the posterior diminishes.</br>
This is because the Bayesian update involves a proportional relationship, and the posterior value approaches saturation.

The Important Concept of Bayesian Reasoning

This diagram illustrates that Bayesian inference:
- Does not discard the prior
- Adds the evidence to it
- Generates a new belief

In other words:
Initial Belief
+
New Evidence
| 
↓
Updated Belief

### Bayesian Update Visualization

This visualization demonstrates how Bayesian reasoning updates beliefs after observing new evidence.

As the prior probability increases, the posterior probability also increases, showing that initial assumptions influence the final belief.

The relationship is nonlinear because Bayesian updating combines prior knowledge with likelihood information through a probabilistic ratio.
