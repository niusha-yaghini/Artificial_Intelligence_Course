<h1> Value Iteration Algorithm 

The main difference from Policy Iteration:

In Policy Iteration:

Policy</br>
↓</br>
Policy Evaluation</br>
↓</br>
Value Function</br>
↓</br>
Policy Improvement

This means we start with a policy and improve it.

However, Value Iteration posits:</br>
There is no need to evaluate the policy separately; find the optimal value directly.

That is:

Value Update</br>
↓</br>
Value Update</br>
↓</br>
Value Update</br>
↓</br>
Optimal Value

The policy is then extracted from the value function.

```python
import gymnasium as gym
import numpy as np

from src.value_iteration import (
    value_iteration,
    extract_policy,
)
```

<h3> Step 1: Environment

```python
env = gym.make(
    "FrozenLake-v1",
    is_slippery=False
)
```

<h3> Step 2: Parameters

```python
gamma = 0.99

n_states = env.observation_space.n

n_actions = env.action_space.n
```

<h3> Step 3: Initialize Value Function

Unlike Policy Iteration, we do not have a policy.

```python
V = np.zeros(
    n_states
)
```

<h3> Step 4: Value Update Function

For each state, we examine all actions and then take the maximum.

```python
optimal_V = value_iteration(
    env,
    n_states,
    n_actions,
    gamma=0.99,
    theta=1e-8
)
```

```python
print(
    optimal_V.reshape(4,4)
)
```

<h3> Step 5: Policy Extraction

```python
optimal_policy = extract_policy(
    env,
    n_states,
    n_actions,
    optimal_V,
    gamma,
)
```

```python
optimal_policy
```

```python
action_map = {
    0:"←",
    1:"↓",
    2:"→",
    3:"↑"
}

for state in range(n_states):
    action = np.argmax(
        optimal_policy[state]
    )

    print(
        state,
        action_map[action]
    )
```

#### Why did Policy Iteration and Value Iteration yield the same result?
Because both are seeking:

$\pi^*$

They differ only in the path taken to reach it.

#### A significant difference in speed

In small environments like FrozenLake, both perform excellently.</br>
However, in large environments, Policy Iteration can be more computationally expensive because each iteration requires:
- Performing a full Policy Evaluation.

In contrast, Value Iteration:
- Performs only a single Bellman update.

### Value Iteration Results

Value Iteration was implemented using the Bellman Optimality Equation.

Unlike Policy Iteration, which alternates between policy evaluation and improvement, Value Iteration directly updates the value function by selecting the maximum expected return over all possible actions.

The algorithm converged to the same optimal value function and policy obtained by Policy Iteration.

The learned policy successfully navigates the FrozenLake environment and reaches the goal while avoiding terminal failure states.
